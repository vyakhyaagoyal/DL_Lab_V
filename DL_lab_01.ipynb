```python
# A one-dimensional NumPy array (vector)
x = np.array([2, 4, 6, 8, 10], dtype=float)

print("x =", x)
print("shape =", x.shape)
print("size =", x.size)
print("ndim =", x.ndim)
print("dtype =", x.dtype)
```

```python
# Two-dimensional array (matrix)
A = np.array([[1, 2, 3],
[4, 5, 6]])

# Row vector and column vector
row_vector = np.array([[1, 2, 3, 4]])
column_vector = np.array([[1], [2], [3], [4]])

print("A:\n", A)
print("A shape:", A.shape)
print("A size:", A.size)
print("A ndim:", A.ndim)
print("Row vector shape:", row_vector.shape)
print("Column vector shape:", column_vector.shape)
```

```python
Z = np.zeros((3, 4)) # zero matrix
O = np.ones((2, 5)) # one matrix
I = np.eye(4) # identity matrix
D = np.diag([2, 4, 6, 8]) # diagonal matrix
C = np.full((2, 3), 7) # constant matrix
R = np.random.randn(3, 2) # standard normal random values
U = np.random.rand(3, 2) # uniform random values in [0, 1)

print("Zero matrix:\n", Z)
print("One matrix:\n", O)
print("Identity matrix:\n", I)
print("Diagonal matrix:\n", D)
print("Constant matrix:\n", C)
print("Random normal matrix:\n", R)
```

```python
M = np.array([[10, 11, 12, 13],
[20, 21, 22, 23],
[30, 31, 32, 33],
[40, 41, 42, 43]])

print("First element:", M[0, 0])
print("Last element:", M[-1, -1])
print("Second row:", M[1, :])
print("Third column:", M[:, 2])
print("Top-left 2 x 2 submatrix:\n", M[0:2, 0:2])
print("Rows 2 to 4, columns 2 to 4:\n", M[1:4, 1:4])
print("Every alternate row:\n", M[::2, :])
```

```python
P = np.array([[1, 2], [4, 5]])
Q = np.array([[10, 20],[40, 50]])
print("P + Q:\n", P + Q)
print("P * Q (element-wise):\n", P * Q)
print("P @ Q (element-wise):\n", P @ Q)
print("P ** 2:\n", P ** 2)
print("Transpose of P:\n", P.T)

# Broadcasting: add one bias value to every row
bias = np.array([[100], [200]])
print("P + bias:\n", P + bias)

# Reshaping
v = np.arange(1, 13)

print("Reshaped 3 x 4:\n", v.reshape(3, 4))
print("Flattened array:", P.flatten())
```

```python
# A row-like 1D vector with 3 elements
v = np.array([1, 2, 3])

# Matrix with shape (3, 2)
N = np.array([[1, 2],
[3, 4],
[5, 6]])

print("v shape:", v.shape)
print("N shape:", N.shape)
print("v @ N shape:", (v @ N).shape)
print("v @ N =", v @ N)

# Matrix-vector product: N.T has shape (2, 3), v has shape (3,)
print("N.T @ v =", N.T @ v)
```

```python
A = np.array([[1, 2, 3],
[4, 5, 6]]) # shape (2, 3)
B = np.array([[1, 0],
[0, 1],
[1, 1]]) # shape (3, 2)

C = A @ B
print("A shape:", A.shape)
print("B shape:", B.shape)
print("C = A @ B:\n", C)
print("C shape:", C.shape)
```

```python
#task 8
#l1 - manhattan distance with order=1
#l2 - euclidean distance with order=2
v = np.array([3, 4, 12], dtype=float)

l2_manual = np.sqrt(np.sum(v ** 2))
l2_numpy = np.linalg.norm(v, ord=2)

print("Manual L2 norm:", l2_manual)
print("NumPy L2 norm:", l2_numpy)

# Unit vector in the same direction
v_unit = v / np.linalg.norm(v)
print("Unit vector:", v_unit)
print("L2 norm of unit vector:", np.linalg.norm(v_unit))
```
